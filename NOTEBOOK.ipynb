## 1. Fullfillment Warehouse Application
The experiment investigates the performance of RHCR in a simulation of the Fulfillment Warehouse, using a 33 × 46 grid map with 16% obstacles.

- Compares the **PBS** performance: runtime and throughput using **RHCR** approach, **HE** (Holding Endpoints) and **RDP** (Reserving Dummy Paths)
- RHCR uses time horizon **w = 20** and replanning period **h = 5**
- HE and RDP replan paths at **every timestamp** basing on Method 3
- **m** indicates number of agents in all experiments
- **5000 timestamps** for every experiment

```python
import pandas as pd
import numpy as np
import re

# Average Runtime Function
def avg_runtime(file, col):
    df = pd.read_csv(file)
    df.columns = col
    runtime_mean = df.runtime.mean()
    runtime_std = df.runtime.std()
    print(f'Average runtime: {runtime_mean:.2f} +/- {runtime_std:.2f}') # Average runtime 
# Throughput Counter Function
def throughput(file, timesteps):
    if timesteps <= 0:
        return 0
    with open(file, "r") as f:
        content = f.read()
    task_counts = [int(match.group(1)) for match in re.finditer(r'(\d+)\s+tasks\s+has\s+been\s+finished', content)]
    total_tasks = sum(task_counts)
    throughput = total_tasks / timesteps
    print(f'Throughput: {throughput}') # Throuhgput
```

```python
# Features names saved durgin the simulation
column_names = [
    "runtime",
    "HL_num_expanded",
    "HL_num_generated",
    "LL_num_expanded",
    "LL_num_generated",
    "solution_cost",
    "min_sum_of_costs",
    "avg_path_length",
    "num_of_collisions",
    "timestep",
    "num_agents",
    "seed",
    "window"
]
# Simulation files
solver_file = 'solver.csv'
file_path = 'Timesteps.txt'
timesteps = 500
```

- **Directed movements** on the map to make MAPF solvers more efficient

- **m** indicates number of agents in all experiments
- **1000 timestamps** for every experiment (reduced number of steps)
- SOC potential funtion with **p = 1** for every experiment
